# Convergence tests with `pw.x`

We run convergence tests in order to make sure we control and understand the extent to which the numerical approximations we are making affect the final outputs of our calculations. In plane-wave DFT two key numerical parameters are:

- the kinetic-energy cutoff `ecutwfc`, which determines how large the set of plane waves used as basis for the Kohn–Sham wavefunctions is
- the k-point mesh, which determines how densely the Brillouin zone is sampled in each direction

The goal is to find the smallest values of these parameters for which the target quantities (total energy, forces) change by less than a chosen threshold — e.g. 5 meV/atom for the energy.

## Optional exercise — k-point convergence via shift difference

In Problem 2 you measured convergence by comparing each mesh against the densest one — but that reference may not be converged itself. Here you get a self-contained error estimate for free, reusing the Problem 2 results as your `sk=0` data and running one new sweep with `sk=1`.

> [!NOTE]
> A uniform grid works because summing any oscillating wave at equally-spaced points causes exact destructive interference — the contributions cancel on the complex unit circle. A Monkhorst-Pack grid is a perfect filter: it annihilates almost every Fourier component of the integrand, leaving only the constant average you want plus tiny high-frequency aliases that vanish as the grid gets denser.

As the mesh gets denser both grids converge to the same energy, so `|E(sk=1) − E(sk=0)|` shrinks monotonically and is a direct measure of how well-converged you are.

**Part A.** Build a list of cases using the same `NK_VALUES` as Problem 2 but passing `sk=1` to `build_mgo_input`. Run the sweep into a new `KSHIFT_DIR`.

**Part B.** You now have two result lists: `k_results` from Problem 2 (`sk=0`) and the new one (`sk=1`). From each extract the energy in Ry and the number of irreducible k-points. Compute `dE_shift_mev_atom` and call `plot_k_shift_convergence`.

`first_globally_converged_index(values, threshold)` returns the first index from which *all subsequent values* stay below the threshold — a strict tail criterion that avoids declaring convergence prematurely.

`plot_k_shift_convergence` takes:
- `NK_VALUES` — the x-axis mesh sizes
- `e_gamma_ev_atom`, `e_shift_ev_atom` — energies in eV/atom for sk=0 (from Problem 2) and sk=1 (new sweep)
- `dE_shift_mev_atom` — `|E(sk=1) − E(sk=0)|` in meV/atom, your error gauge
- `nk_irr_gamma`, `nk_irr_shift` — irreducible k-point counts for each shift
- `thr_mev_atom` — the convergence threshold drawn as a horizontal line

> [!TIP]
> To build `dE_shift_mev_atom`, zip the two energy lists (in Ry) and take the absolute difference for each pair, then convert: multiply by `RY_TO_EV / nat * 1000`.

The plot has three panels: both energies converging toward each other, the shift difference decaying, and the irreducible k-point counts — you may notice the shifted grid has *fewer* irreducible k-points because the Γ point carries extra symmetry that inflates the sk=0 count.
